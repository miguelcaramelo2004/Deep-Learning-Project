```python
import os
import torch
import torch.nn as nn
import timm
import numpy as np
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, Subset
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score
from torch.amp import GradScaler, autocast
from tqdm.auto import tqdm

```

```python
# --- 1. CONFIGURATION ---
DATA_PATH = "dataset/wikiart"
MODEL_NAME = 'convnext_base.fb_in22k_ft_in1k_384'
IMG_SIZE = 384
BATCH_SIZE = 8 
ACCUMULATION_STEPS = 8 
MAX_EPOCHS = 40
WARMUP_EPOCHS = 3
PATIENCE = 6 
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

For **training**, it employs aggressive data augmentation techniques—such as random cropping, horizontal flipping, grayscale conversion, and sharpness adjustments—to enhance the model's robustness and prevent overfitting. 

For **evaluation**, it uses a deterministic approach, resizing and center-cropping the image to ensure consistent input. 

Both pipelines conclude by normalizing the data using standard ImageNet statistics to stabilize the neural network's learning process.

```python
# --- 2. TRANSFORMS (Selective Texture + Microscope) ---
def get_transforms(img_size, is_training=True):
    if not is_training:
        return transforms.Compose([
            transforms.Resize(img_size + 32),
            transforms.CenterCrop(img_size),
            transforms.ToTensor(),
            transforms.Normalize(timm.data.IMAGENET_DEFAULT_MEAN, timm.data.IMAGENET_DEFAULT_STD)
        ])
    return transforms.Compose([
        transforms.Resize(512),
        transforms.RandomResizedCrop(img_size, scale=(0.08, 1.0)),
        transforms.RandomHorizontalFlip(),
        transforms.RandomGrayscale(p=0.2), 
        transforms.RandomAdjustSharpness(2, p=0.5),
        transforms.ToTensor(),
        transforms.Normalize(timm.data.IMAGENET_DEFAULT_MEAN, timm.data.IMAGENET_DEFAULT_STD)
    ])

```

Implementing two advanced training strategies: 
- **Layer-wise Learning Rate Decay (LLRD)**: `get_llrd_optimizer` function assigns progressively lower learning rates to earlier layers of the model, ensuring that foundational features remain stable while deeper layers adapt more specifically to the task. 
- **Mixup augmentation**: `apply_mixup` function performs linear interpolation between pairs of input images and their labels, forcing the model to learn smoother decision boundaries and significantly reducing overfitting.

```python
# --- 3. SOTA UTILS (LLRD, Mixup, IDF Weights) ---
def get_llrd_optimizer(model, lr, layer_decay=0.85):
    n_stages = len(model.stages)
    params = []
    # Stem
    stem_lr = lr * (layer_decay ** (n_stages + 1))
    params.append({'params': model.stem.parameters(), 'lr': stem_lr, 'initial_lr': stem_lr})
    # Stages
    for i, stage in enumerate(model.stages):
        stage_lr = lr * (layer_decay ** (n_stages - i))
        params.append({'params': stage.parameters(), 'lr': stage_lr, 'initial_lr': stage_lr})
    # Head
    params.append({'params': model.head.parameters(), 'lr': lr, 'initial_lr': lr})
    return torch.optim.AdamW(params, weight_decay=0.05)

def apply_mixup(x, y, alpha=0.8):
    lam = np.random.beta(alpha, alpha)
    index = torch.randperm(x.size(0)).to(DEVICE)
    mixed_x = lam * x + (1 - lam) * x[index]
    return mixed_x, y, y[index], lam

```

By utilizing an `IDF-style weighting scheme`, we assign higher importance to underrepresented classes, directly optimizing the model for Macro-F1 performance rather than simple accuracy. 

Additionally, the implementation employs stratified splitting for the validation set, ensuring that the distribution of labels remains consistent across both training and evaluation subsets.

```python
# --- 4. DATA PREPARATION ---
full_ds = datasets.ImageFolder(DATA_PATH)
num_classes = len(full_ds.classes)
class_counts = np.bincount(full_ds.targets)
# IDF-style weighting for Macro-F1 focus
weights = torch.FloatTensor(np.log(len(full_ds) / class_counts)).to(DEVICE)

train_idx, val_idx = train_test_split(
    list(range(len(full_ds))), test_size=0.15, stratify=full_ds.targets, random_state=42
)
```

This initialization phase sets up a high-performance training environment using the *timm library* for transfer learning. 

The model incorporates **Stochastic Depth (via drop_path_rate)** and **Label Smoothing** to prevent overfitting and improve robustness. 

The loss function is explicitly weighted to handle class imbalance, while the *GradScaler* enables **Mixed Precision training**, significantly optimizing GPU memory usage and computational speed without sacrificing model accuracy.

```python

# --- 5. INITIALIZATION ---
try:
    model = timm.create_model(MODEL_NAME, pretrained=True, num_classes=num_classes, drop_path_rate=0.2).to(DEVICE)
except Exception as e:
    print(f"Error loading model: {e}. Checking timm installation...")
    raise

optimizer = get_llrd_optimizer(model, 4e-5)
criterion = nn.CrossEntropyLoss(weight=weights, label_smoothing=0.1)
scaler = GradScaler('cuda')
```

The `validate function` serves as the model's performance benchmark, executed in evaluation mode to ensure deterministic behavior. 

By utilizing *torch.no_grad()* and *autocast*, it efficiently computes predictions while minimizing computational overhead. 

The function returns a comprehensive set of metrics, including loss, accuracy, and Macro-F1 score. 

```python
# --- 6. TRAINING & EVALUATION FUNCTIONS ---
def validate(model, loader, device, current_criterion):
    model.eval()
    val_loss, all_preds, all_labels = 0, [], []
    with torch.no_grad():
        for imgs, labels in loader:
            imgs, labels = imgs.to(device), labels.to(device)
            with autocast('cuda'):
                outputs = model(imgs)
                loss = current_criterion(outputs, labels)
            val_loss += loss.item()
            all_preds.extend(torch.argmax(outputs, dim=1).cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
    
    return val_loss / len(loader), accuracy_score(all_labels, all_preds), f1_score(all_labels, all_preds, average='macro')
```

```python

# --- 7. THE MAIN LOOP ---
history = {"train_loss": [], "val_loss": [], "val_acc": [], "val_f1": []}
best_f1, early_stop_counter, has_surged = 0, 0, False

for epoch in range(MAX_EPOCHS):
    # FixRes Logic: Once triggered, remains active
    if not has_surged:
        if epoch >= (MAX_EPOCHS - 6) or early_stop_counter >= 3:
            has_surged = True
            print("[PERMANENT SURGE] Calibrating at 512px FixRes mode.")

    current_res = 512 if has_surged else 384
    is_fixres = has_surged
    
    train_loader = DataLoader(Subset(datasets.ImageFolder(DATA_PATH, get_transforms(current_res, not is_fixres)), train_idx), 
                              batch_size=BATCH_SIZE, shuffle=True, num_workers=4, pin_memory=True)
    val_loader = DataLoader(Subset(datasets.ImageFolder(DATA_PATH, get_transforms(current_res, False)), val_idx), 
                            batch_size=BATCH_SIZE, shuffle=False, num_workers=4)

    model.train()
    running_loss, train_correct, train_total = 0.0, 0, 0
    optimizer.zero_grad()
    
    # Warmup Logic
    if epoch < WARMUP_EPOCHS:
        warmup_factor = (epoch + 1) / WARMUP_EPOCHS
        for pg in optimizer.param_groups: pg['lr'] = pg['initial_lr'] * warmup_factor

    pbar = tqdm(train_loader, desc=f"Ep {epoch+1} [{current_res}px]")
    for i, (imgs, labels) in enumerate(pbar):
        imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
        
        with autocast('cuda'):
            if not is_fixres and np.random.rand() > 0.5:
                mixed_x, y_a, y_b, lam = apply_mixup(imgs, labels)
                outputs = model(mixed_x)
                loss = (lam * nn.CrossEntropyLoss()(outputs, y_a) + (1 - lam) * nn.CrossEntropyLoss()(outputs, y_b))
            else:
                outputs = model(imgs)
                loss = nn.CrossEntropyLoss(label_smoothing=0.1)(outputs, labels)
            
            loss_scaled = loss / ACCUMULATION_STEPS
        
        scaler.scale(loss_scaled).backward()
        
        if (i + 1) % ACCUMULATION_STEPS == 0:
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad()

        # Tracking (using argmax on final outputs for acc)
        running_loss += loss.item()
        _, predicted = outputs.max(1)
        train_total += labels.size(0)
        train_correct += predicted.eq(labels).sum().item()
        pbar.set_postfix({"L": f"{running_loss/(i+1):.3f}", "A": f"{100.*train_correct/train_total:.1f}%"})

    # Validate
    v_loss, v_acc, v_f1 = validate(model, val_loader, DEVICE, nn.CrossEntropyLoss())
    
    history["train_loss"].append(running_loss / len(train_loader))
    history["val_loss"].append(v_loss)
    history["val_acc"].append(v_acc)
    history["val_f1"].append(v_f1)

    print(f"Summary Ep {epoch+1}: Val Loss: {v_loss:.3f} | Val Acc: {v_acc:.4f} | Val F1: {v_f1:.4f}")

    if v_f1 > best_f1:
        best_f1 = v_f1
        torch.save(model.state_dict(), "wikiart_sota_best.pth")
        early_stop_counter = 0
    else:
        early_stop_counter += 1
        if early_stop_counter >= PATIENCE:
            print(f"Early Stopping at Ep {epoch+1}. Best F1: {best_f1:.4f}"); break
```

```python

# --- 8. FINAL SAVE ---
torch.save(model.state_dict(), "final_model_weights.pth")
print(f"Training Complete. SOTA Best F1: {best_f1:.4f}")
```
